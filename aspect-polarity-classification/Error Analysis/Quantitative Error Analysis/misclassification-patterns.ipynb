```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

labels = ["Positive", "Neutral", "Negative"]

datasets = {
    "Restaurant15": np.array([
        [0, 9, 18],
        [52, 0, 21],
        [6, 21, 0]
    ]),
    "Laptop15": np.array([
        [0, 15, 24],
        [25, 0, 50],
        [2, 6, 0]
    ]),
    "FABSA": np.array([
        [0, 36, 6],
        [5, 0, 11],
        [61, 6, 0]
    ])
}

# Replace with actual class counts
class_counts = {
    "Restaurant15": [728, 196, 196],
    "Laptop15": [341, 169, 128],
    "FABSA": [1825, 91, 896]
}

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, (name, cm) in zip(axes, datasets.items()):

    counts = class_counts[name]

    annot = np.empty(cm.shape, dtype=object)

    for i in range(3):
        for j in range(3):

            if i == j:
                annot[i, j] = "--"
            else:
                pct = 100 * cm[i, j] / counts[i]
                annot[i, j] = f"{cm[i,j]}\n({pct:.1f}%)"

    sns.heatmap(
        cm,
        annot=annot,
        fmt="",
        cmap="Blues",
        xticklabels=labels,
        yticklabels=labels,
        cbar=False,
        linewidths=0.5,
        square=True,
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted Label")
    ax.set_ylabel("True Label")

plt.savefig("apc_confusion_matrix.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```
