```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from pyabsa import AspectTermExtraction as ATEPC
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
import numpy as np
import json
```

```python
def smart_load_and_parse(data_input):
    # 1. Handle Pandas DataFrame
    if isinstance(data_input, pd.DataFrame):
        print("Detected Pandas DataFrame. Converting to list of dicts...")
        # orient='records' turns it into [{'text':..., 'aspects':...}, ...]
        data_input = data_input.to_dict(orient='records')

    # 2. Handle String (JSON string)
    elif isinstance(data_input, str):
        print("Detected String. Parsing as JSON...")
        try:
            data_input = json.loads(data_input)
        except:
            # Maybe it's a file path?
            try:
                with open(data_input, 'r') as f:
                    data_input = json.load(f)
            except:
                raise ValueError("Could not parse input string/path.")

    # 3. Handle Dictionary (Root key wrapper)
    # If data is a dict (not a list), it might be {"train": [...]} or {"data": [...]}
    if isinstance(data_input, dict):
        print(f"Detected Dictionary with keys: {list(data_input.keys())}")
        # Try to find the list inside common keys
        for key in ['data', 'train', 'test', 'reviews', 'sentence']:
            if key in data_input and isinstance(data_input[key], list):
                print(f"Found list under key: '{key}'")
                data_input = data_input[key]
                break
        else:
            # If no common key found, maybe it's a single entry? Wrap it.
            if 'text' in data_input:
                data_input = [data_input]
            else:
                raise ValueError("Input is a dictionary but couldn't find a list of data inside.")

    # Now verify we have a list
    if not isinstance(data_input, list):
        raise TypeError(f"Expected a list of data, but got {type(data_input)}")

    return parse_custom_dataset(data_input)
```

```python
def parse_custom_dataset(data):
    """
    Parses dataset where 'aspects' is a dictionary of lists.
    
    Format expected:
    {
      "text": "Sentence here",
      "aspects": {
          "term": ["aspect1", "aspect2"],
          "polarity": ["positive", "negative"],
          "from": [10, 25],
          "to": [17, 32]
      }
    }
    """
    cleaned_data = []
    
    for entry in data:
        text = entry['text']
        ground_truth_terms = []
        
        aspect_data = entry['aspects']
        
        # Scenario A: The 'term' list exists (e.g., ["staff", "food"])
        if 'term' in aspect_data and isinstance(aspect_data['term'], list):
            ground_truth_terms = aspect_data['term']
            
        # Scenario B: 'term' is missing, but 'from'/'to' offsets exist
        elif 'from' in aspect_data and 'to' in aspect_data:
            starts = aspect_data['from']
            ends = aspect_data['to']
            # Slice text using offsets
            for s, e in zip(starts, ends):
                ground_truth_terms.append(text[int(s):int(e)])
        
        cleaned_data.append({
            "text": text,
            "ground_truth": ground_truth_terms
        })
        
    return cleaned_data
```

```python
def calculate_metrics(gold_data, pred_data):
    tp, fp, fn = 0, 0, 0
    exact_matches = 0
    per_sample_accuracies = []

    for gold_list, pred_list in zip(gold_data, pred_data):
        gold_set = set([str(g).lower().strip() for g in gold_list])
        pred_set = set([str(p).lower().strip() for p in pred_list])

        correct = len(gold_set & pred_set)
        total = len(gold_set)

        # Per-sample accuracy (e.g., 3/4 → 0.75)
        if total > 0:
            per_sample_accuracies.append(correct / total)
        else:
            per_sample_accuracies.append(0)

        tp += correct
        fp += len(pred_set - gold_set)
        fn += len(gold_set - pred_set)

        if gold_set == pred_set:
            exact_matches += 1

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    accuracy_em = exact_matches / len(gold_data) if len(gold_data) > 0 else 0
    accuracy_overall = sum(per_sample_accuracies) / len(per_sample_accuracies)

    return {
        "Precision": round(precision * 100, 2),
        "Recall": round(recall * 100, 2),
        "F1 Score": round(f1 * 100, 2),
        "Accuracy (Overall)": round(accuracy_overall * 100, 2),
        "Accuracy (Exact Match)": round(accuracy_em * 100, 2)
    }
```

```python
def evaluate_pyabsa(dataset):
    print("\n--- Evaluating PyABSA ---")
    # Load multilingual or english checkpoint
    aspect_extractor = ATEPC.AspectExtractor('english', auto_device=True)
    
    preds = []
    golds = [d['ground_truth'] for d in dataset]
    
    for i, item in enumerate(dataset):
        try:
            result = aspect_extractor.predict(item['text'], print_result=False)
            preds.append(result['aspect']) # Extract predicted list
        except Exception as e:
            print(f"Error in PyABSA prediction: {e}")
            preds.append([])

        if i % 10 == 0: print(f"PyABSA: Processed {i}/{len(dataset)}...")

    return calculate_metrics(golds, preds)
```

```python
def evaluate_instructabsa(dataset, version=2):
    print(f"\n--- Evaluating InstructABSA V{version} ---")
    
    device = "cuda" if torch.cuda.is_available() else "cpu"
    
    # 1. Select Model & Prompt based on Version
    if version == 2:
        # InstructABSA-2 (Uses Pos, Neg, Neutral examples) - Best Perf
        checkpoint = "kevinscaria/ate_tk-instruct-base-def-pos-neg-neut-combined"
        prompt_prefix = (
            "Definition: The task is to extract the aspect terms from the given sentence. "
            "Example 1: The pasta was delicious. Output: pasta "
            "Example 2: The service was friendly. Output: service "
            "Example 3: The ambience was perfect for a date night. Output: ambience "
            "Example 4: The prices are reasonable. Output: prices "
            "Sentence: "
        )
    else:
        # InstructABSA-1 (Uses only Positive examples) - Specific to Restaurants often
        checkpoint = "kevinscaria/ate_tk-instruct-base-def-pos-restaurants"
        prompt_prefix = (
            "Definition: The task is to extract all aspect terms from the given sentence. "
            "Example 1: The food was great. Output: food "
            "Example 2: The service was slow. Output: service "
            "Example 3: The ambience was okay. Output: ambience "
            "Example 4: The waiter didn't respond for 10 minutes. Output: waiter "
            "Example 5: Nothing special about the desserts. Output: desserts "
            "Sentence: "
        )

    tokenizer = AutoTokenizer.from_pretrained(checkpoint)
    model = AutoModelForSeq2SeqLM.from_pretrained(checkpoint).to(device)
    
    preds = []
    golds = [d['ground_truth'] for d in dataset]

    for i, item in enumerate(dataset):
        input_text = prompt_prefix + item['text']
        input_ids = tokenizer(input_text, return_tensors="pt").input_ids.to(device)
        
        with torch.no_grad():
            outputs = model.generate(input_ids, max_length=128)
        
        decoded = tokenizer.decode(outputs[0], skip_special_tokens=True)
        
        # Parse output "aspect1 | aspect2"
        extracted = [t.strip() for t in decoded.split('|') if t.strip()]
        preds.append(extracted)
        
        if i % 10 == 0: print(f"InstructABSA{version}: Processed {i}/{len(dataset)}...")

    return calculate_metrics(golds, preds)
```

```python
my_data = my_data.to_list()
```

```python
print("Parsing dataset...")
processed_data = smart_load_and_parse(my_data)
print(f"Loaded {len(processed_data)} samples.")
print(f"Sample 1 Ground Truth: {processed_data[0]['ground_truth']}")
```

```python
# 2. Run Evaluations
results_pyabsa = evaluate_pyabsa(processed_data)
results_instruct1 = evaluate_instructabsa(processed_data, version=1)
results_instruct2 = evaluate_instructabsa(processed_data, version=2)
```
