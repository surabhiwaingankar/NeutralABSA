```python
import json
import torch
import numpy as np
from torch.utils.data import DataLoader, TensorDataset
from transformers import AutoTokenizer
from transformers import RobertaForTokenClassification
from sklearn.metrics import accuracy_score, classification_report
from peft import LoraConfig, get_peft_model, TaskType
from datasets import load_dataset
```

```python
# Load dataset
dataset_restaurant = load_dataset("jakartaresearch/semeval-absa", name='laptop')
train_ds = dataset_restaurant["train"]
test_ds = dataset_restaurant["validation"]
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
# Label map
label_map = {"O": 0, "B": 1, "I": 2}
id2label = {v: k for k, v in label_map.items()}
```

```python
#load bert model
model = RobertaForTokenClassification.from_pretrained("roberta-base", num_labels=len(label_map))
```

```python
#Load bert tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "roberta-base",
    use_fast=True   # Force Fast tokenizer
)
```

```python
# Apply PEFT with LoRA
lora_config = LoraConfig(
    r=8,  # Rank of the LoRA matrix
    lora_alpha=16,  # Scaling factor
    lora_dropout=0.1,
    bias="none",
    task_type=TaskType.TOKEN_CLS  # Assuming token classification task
)
peft_model = get_peft_model(model, lora_config)
peft_model.to(device)
peft_model.train()
```

```python
# Tokenize
def tokenize_and_align(ex):
    text = ex["text"]
    terms = ex["aspects"]["term"]
    from_idx = ex["aspects"]["from"]
    to_idx = ex["aspects"]["to"]

    encoding = tokenizer(
        text,
        return_offsets_mapping=True,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )

    labels = ["O"] * len(encoding["offset_mapping"][0])

    for start, end in zip(from_idx, to_idx):
        span_token_indices = []

        # collect all token positions belonging to this aspect span
        for i, (s, e) in enumerate(encoding["offset_mapping"][0]):
            if s == 0 and e == 0:   # padding tokens
                continue
            if s >= start and e <= end:
                span_token_indices.append(i)

        # assign BIO tags
        if len(span_token_indices) > 0:
            labels[span_token_indices[0]] = "B"       # first token
            for idx in span_token_indices[1:]:        # remaining tokens
                labels[idx] = "I"

    label_ids = [label_map[l] for l in labels]

    return (
        encoding["input_ids"].squeeze(),
        encoding["attention_mask"].squeeze(),
        torch.tensor(label_ids)
    )
```

```python
# Build tensors
def prepare(ds):
    input_ids, attention_masks, label_ids = [], [], []
    for ex in ds:
        a, b, c = tokenize_and_align(ex)
        input_ids.append(a)
        attention_masks.append(b)
        label_ids.append(c)
    return TensorDataset(torch.stack(input_ids), torch.stack(attention_masks), torch.stack(label_ids))

```

```python
train_dataset = prepare(train_ds)
test_dataset = prepare(test_ds)
```

```python
train_loader = DataLoader(train_dataset, batch_size=2, shuffle=True)
optimizer = torch.optim.AdamW(peft_model.parameters(), lr=5e-5)
```

```python
# Training loop
for epoch in range(3):
    for batch in train_loader:
        input_ids, attention_mask, labels = [x.to(device) for x in batch]
        outputs = peft_model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
    print(f"Epoch {epoch + 1} loss: {loss.item()}")
```

```python
import torch
import re
import string
from rapidfuzz import fuzz
```

```python
def bio_to_spans(tokens, labels):
    """
    Convert BIO token labels to aspect spans (strings).
    tokens: list of string tokens from tokenizer
    labels: list of BIO labels (B/I/O)
    """
    spans = []
    current_span = []

    for tok, lab in zip(tokens, labels):
        if lab == "B":
            # finish previous span
            if current_span:
                spans.append(" ".join(current_span))
                current_span = []

            # start new span
            tok = tok.replace("Ġ", "").strip()
            if tok:
                current_span.append(tok)

        elif lab == "I":
            tok = tok.replace("Ġ", "").strip()
            if tok:
                current_span.append(tok)

        else:  # "O"
            if current_span:
                spans.append(" ".join(current_span))
                current_span = []

    # end of sentence case
    if current_span:
        spans.append(" ".join(current_span))

    return spans
```

```python
def normalize_span(x):
    """Clean and normalize predicted or gold span text"""
    x = re.sub(r"(^Ġ+|\s+Ġ+)", " ", x)
    x = x.lower().strip()
    x = x.translate(str.maketrans("", "", string.punctuation.replace("-", "").replace("/", "")))
    x = re.sub(r"\s+", " ", x)
    edge_stop = ["the", "a", "an", "and", "or", "for", "of", "to", "with", "is", "was", "were"]
    tokens = x.split()
    while tokens and tokens[0] in edge_stop:
        tokens.pop(0)
    while tokens and tokens[-1] in edge_stop:
        tokens.pop()
    return " ".join(tokens).strip()
```

```python
def remove_subspans(spans):
    """Keep only the longest spans in case of overlapping / sub-spans"""
    spans = sorted(spans, key=len, reverse=True)
    filtered = []
    for s in spans:
        if not any(s in f for f in filtered):
            filtered.append(s)
    return filtered
```

```python
def calculate_span_metrics_fuzzy(pred_spans_list, gold_spans_list, threshold=80):
    """
    Compute span-level metrics using fuzzy/partial matching.
    threshold: % similarity for counting as correct
    """
    tp, fp, fn = 0, 0, 0
    exact_matches = 0
    per_sample_accuracies = []

    for gold, pred in zip(gold_spans_list, pred_spans_list):
        gold_set = set([normalize_span(g) for g in gold])
        pred_set = set([normalize_span(p) for p in pred])

        gold_set = set(remove_subspans(list(gold_set)))
        pred_set = set(remove_subspans(list(pred_set)))

        matched_pred = set()
        matched_gold = set()

        # Fuzzy matching
        for g in gold_set:
            for p in pred_set:
                if fuzz.ratio(g, p) >= threshold:
                    tp += 1
                    matched_pred.add(p)
                    matched_gold.add(g)
                    break

        fp += len(pred_set - matched_pred)
        fn += len(gold_set - matched_gold)

        # Per-sample accuracy
        correct = len(matched_gold)
        total = len(gold_set)
        per_sample_accuracies.append(correct / total if total > 0 else 0)

        if matched_gold == gold_set and matched_pred == pred_set:
            exact_matches += 1

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
    overall_acc = sum(per_sample_accuracies) / len(per_sample_accuracies)
    exact_acc = exact_matches / len(pred_spans_list) if pred_spans_list else 0

    return {
        "Precision": round(precision * 100, 2),
        "Recall": round(recall * 100, 2),
        "F1 Score": round(f1 * 100, 2),
        "Accuracy (Overall)": round(overall_acc * 100, 2),
        "Accuracy (Exact Match)": round(exact_acc * 100, 2)
    }
```

```python
# --- Run evaluation ---
pred_spans_list = []
gold_spans_list = []

peft_model.eval()
with torch.no_grad():
    for ex in test_ds:
        text = ex["text"]
        gold_spans = ex["aspects"]["term"]

        encoding = tokenizer(text, return_tensors="pt", truncation=True).to(device)
        logits = peft_model(**encoding).logits
        preds = torch.argmax(logits, dim=2)[0].cpu().numpy()

        tokens = tokenizer.convert_ids_to_tokens(encoding["input_ids"][0])
        labels = [id2label[p] for p in preds]

        pred_spans = bio_to_spans(tokens, labels)

        pred_spans_list.append(pred_spans)
        gold_spans_list.append(gold_spans)


# --- Compute metrics with fuzzy matching ---
metrics = calculate_span_metrics_fuzzy(pred_spans_list, gold_spans_list, threshold=80)
print("Span-level Metrics with Fuzzy Matching:")
for k, v in metrics.items():
    print(f"{k}: {v}%")
```
